```python
import os
import sqlite3
import pandas as pd
```

```python
# Create (or connect to) SQLite database file
conn = sqlite3.connect("lending_club_sample_10pct.db")

# Always enable foreign keys in SQLite
conn.execute("PRAGMA foreign_keys = ON;")

cursor = conn.cursor()

print("Database created and connected successfully.")
```

```python
# Create Loan table
cursor.execute("""
CREATE TABLE IF NOT EXISTS loan (
    loan_id TEXT PRIMARY KEY,
    loan_status TEXT,
    loan_amnt REAL,
    term TEXT,
    int_rate REAL,
    installment REAL,
    purpose TEXT,
    grade TEXT,
    sub_grade TEXT,
    issue_d TEXT,
    addr_state TEXT
);
""")

# Create Borrower table
cursor.execute("""
CREATE TABLE IF NOT EXISTS borrower (
    loan_id TEXT PRIMARY KEY,
    annual_inc REAL,
    emp_length TEXT,
    home_ownership TEXT,
    verification_status TEXT,
    FOREIGN KEY (loan_id) REFERENCES loan(loan_id)
);
""")

# Create Credit History table
cursor.execute("""
CREATE TABLE IF NOT EXISTS credit_history (
    credit_history_id INTEGER PRIMARY KEY AUTOINCREMENT,
    loan_id TEXT,
    fico_range_low INTEGER,
    fico_range_high INTEGER,
    earliest_cr_line TEXT,
    open_acc INTEGER,
    total_acc INTEGER,
    delinq_2yrs INTEGER,
    pub_rec INTEGER,
    FOREIGN KEY (loan_id) REFERENCES loan(loan_id)
);
""")

# Create Debt Profile table
cursor.execute("""
CREATE TABLE IF NOT EXISTS debt_profile (
    debt_profile_id INTEGER PRIMARY KEY AUTOINCREMENT,
    loan_id TEXT,
    dti REAL,
    revol_bal REAL,
    revol_util REAL,
    FOREIGN KEY (loan_id) REFERENCES loan(loan_id)
);
""")

conn.commit()
print("Tables created successfully.")
```

```python
import pandas as pd

# Load your cleaned 10% sample
df = pd.read_csv("lending_club_sample_10pct.csv")

# Rename id → loan_id if needed
df = df.rename(columns={"id": "loan_id"})

# Insert into Loan table
loan_cols = [
    "loan_id", "loan_status", "loan_amnt", "term",
    "int_rate", "installment", "purpose",
    "grade", "sub_grade", "issue_d", "addr_state"
]

df[loan_cols].to_sql(
    "loan",
    conn,
    if_exists="append",
    index=False
)

# Insert into Borrower table
borrower_cols = [
    "loan_id", "annual_inc", "emp_length",
    "home_ownership", "verification_status"
]

df[borrower_cols].to_sql(
    "borrower",
    conn,
    if_exists="append",
    index=False
)

# Insert into Credit History table
credit_cols = [
    "loan_id", "fico_range_low", "fico_range_high",
    "earliest_cr_line", "open_acc", "total_acc",
    "delinq_2yrs", "pub_rec"
]

df[credit_cols].to_sql(
    "credit_history",
    conn,
    if_exists="append",
    index=False
)

# Insert into Debt Profile table
debt_cols = ["loan_id", "dti", "revol_bal", "revol_util"]

df[debt_cols].to_sql(
    "debt_profile",
    conn,
    if_exists="append",
    index=False
)

print("Data inserted successfully.")
```

## Visualizations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# === CONFIG ===
CSV_PATH = "lending_club_sample_10pct.csv"   # update path if needed
SAVE_FIGS = False                            # set True to save PNG files
OUT_DIR = "."                                # where to save figs

# === LOAD ===
df = pd.read_csv(CSV_PATH, low_memory=False)

# === CREATE BINARY TARGET ===
default_map = {
    "Fully Paid": 0,
    "Does not meet the credit policy. Status:Fully Paid": 0,
    "Charged Off": 1,
    "Default": 1,
    "Does not meet the credit policy. Status:Charged Off": 1
}
df = df[df["loan_status"].isin(default_map.keys())].copy()
df["default"] = df["loan_status"].map(default_map)

# helper to convert numeric safely
def to_numeric_safe(series):
    return pd.to_numeric(series, errors="coerce")

# helper to save figure
def save(fig_name):
    if SAVE_FIGS:
        plt.savefig(f"{OUT_DIR}/{fig_name}", dpi=150, bbox_inches="tight")
```

```python
# 1) FICO (avg) vs Default rate

if {"fico_range_low", "fico_range_high"}.issubset(df.columns):
    df["fico_avg"] = (to_numeric_safe(df["fico_range_low"]) + to_numeric_safe(df["fico_range_high"])) / 2
    fico = df[["fico_avg", "default"]].dropna()
    fico["fico_bin"] = pd.cut(fico["fico_avg"], bins=np.arange(300, 861, 10))
    fico_grp = fico.groupby("fico_bin")["default"].agg(["mean", "count"]).reset_index()
    fico_grp = fico_grp[fico_grp["count"] >= 50]  # keep bins with at least 50 samples

    plt.figure(figsize=(9,5))
    plt.plot(fico_grp["fico_bin"].astype(str), fico_grp["mean"], marker="o")
    plt.xticks(rotation=45, fontsize=8)
    plt.xlabel("FICO bin (10-point)")
    plt.ylabel("Default rate")
    plt.title("FICO (average) vs Default rate")
    plt.tight_layout()
    save("fico_vs_default.png")
    plt.show()
```

```python
# 2) DTI vs Default rate

if "dti" in df.columns:
    df["dti_num"] = to_numeric_safe(df["dti"])
    dti = df[["dti_num", "default"]].dropna()
    # bins: 0-50 (2-point bins), then large buckets for extreme values
    dti_bins = np.concatenate((np.arange(0, 50, 2), [100, 500]))
    dti["dti_bin"] = pd.cut(dti["dti_num"], bins=dti_bins)
    dti_grp = dti.groupby("dti_bin")["default"].agg(["mean", "count"]).reset_index()
    dti_grp = dti_grp[dti_grp["count"] >= 50]

    plt.figure(figsize=(9,5))
    plt.plot(dti_grp["dti_bin"].astype(str), dti_grp["mean"], marker="o")
    plt.xticks(rotation=45, fontsize=8)
    plt.xlabel("DTI bin")
    plt.ylabel("Default rate")
    plt.title("DTI vs Default rate")
    plt.tight_layout()
    save("dti_vs_default.png")
    plt.show()
```

```python
# 3) Loan Term vs Default rate

if "term" in df.columns:
    term_grp = df.groupby("term")["default"].agg(["mean", "count"]).reset_index().sort_values("term")
    plt.figure(figsize=(6,4))
    plt.bar(term_grp["term"].astype(str), term_grp["mean"])
    plt.xlabel("Loan term")
    plt.ylabel("Default rate")
    plt.title("Default rate by Loan Term")
    for i, v in enumerate(term_grp["mean"]):
        plt.text(i, v + 0.01, f"{v:.2%}", ha="center", fontsize=8)
    plt.tight_layout()
    save("term_vs_default.png")
    plt.show()
```

```python
# 4) Interest Rate vs Default (scatter + binned avg)

if "int_rate" in df.columns:
    df["int_rate_num"] = to_numeric_safe(df["int_rate"])
    ir = df[["int_rate_num", "default"]].dropna()
    plt.figure(figsize=(9,5))
    # scatter a sample for visibility
    sample_ir = ir.sample(n=min(len(ir), 5000), random_state=42)
    plt.scatter(sample_ir["int_rate_num"], sample_ir["default"], alpha=0.02, s=8)
    # binned averages
    bins = np.linspace(ir["int_rate_num"].min(), ir["int_rate_num"].max(), 20)
    ir["ir_bin"] = pd.cut(ir["int_rate_num"], bins=bins)
    ir_grp = ir.groupby("ir_bin")["default"].agg(["mean", "count"]).reset_index()
    ir_grp = ir_grp[ir_grp["count"] >= 30]
    bin_centers = [interval.left + (interval.right-interval.left)/2 for interval in ir_grp["ir_bin"]]
    plt.plot(bin_centers, ir_grp["mean"], marker="o")
    plt.xlabel("Interest rate")
    plt.ylabel("Default rate")
    plt.title("Interest Rate vs Default (scatter + binned average)")
    plt.tight_layout()
    save("int_rate_vs_default.png")
    plt.show()
```

```python
# 5) Annual Income vs Default (log-binned)

if "annual_inc" in df.columns:
    df["annual_inc_num"] = to_numeric_safe(df["annual_inc"])
    inc = df[["annual_inc_num", "default"]].dropna()
    inc = inc[inc["annual_inc_num"] > 0]  # remove non-positive incomes
    inc["log_inc"] = np.log10(inc["annual_inc_num"])
    inc["inc_bin"] = pd.cut(inc["log_inc"], bins=np.linspace(inc["log_inc"].min(), inc["log_inc"].max(), 15))
    inc_grp = inc.groupby("inc_bin")["default"].agg(["mean", "count"]).reset_index()
    inc_grp = inc_grp[inc_grp["count"] >= 30]
    plt.figure(figsize=(9,5))
    plt.plot(inc_grp["inc_bin"].astype(str), inc_grp["mean"], marker="o")
    plt.xticks(rotation=45, fontsize=8)
    plt.xlabel("Log10(Annual Income) bin")
    plt.ylabel("Default rate")
    plt.title("Annual Income (log-binned) vs Default rate")
    plt.tight_layout()
    save("income_vs_default.png")
    plt.show()
```

```python
# 6) Grade vs Default rate (ordinal)

if "grade" in df.columns:
    # ensure ordinal order A..G
    grade_order = sorted(df["grade"].dropna().unique())
    grade_grp = df.groupby("grade")["default"].agg(["mean", "count"]).reindex(grade_order).reset_index()
    plt.figure(figsize=(7,4))
    plt.bar(grade_grp["grade"], grade_grp["mean"])
    plt.xlabel("Grade")
    plt.ylabel("Default rate")
    plt.title("Default rate by Grade (A -> G)")
    for i, v in enumerate(grade_grp["mean"]):
        plt.text(i, v + 0.01, f"{v:.2%}", ha="center", fontsize=9)
    plt.tight_layout()
    save("grade_vs_default.png")
    plt.show()
```
