```python
default_map = {
    "Fully Paid": 0,
    "Does not meet the credit policy. Status:Fully Paid": 0,
    "Charged Off": 1,
    "Default": 1,
    "Does not meet the credit policy. Status:Charged Off": 1
}

# filter to finalized outcomes and create binary target
df = df[df["loan_status"].isin(default_map.keys())].copy()
df["default"] = df["loan_status"].map(default_map)

# drop rows with missing target (should be none after filter but safe)
df = df.dropna(subset=["default"])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

split = StratifiedShuffleSplit(n_splits=1, test_size=0.10, random_state=42)

for _, sample_idx in split.split(df, df["default"]):
    df_sample = df.iloc[sample_idx].copy()
```

```python
# default rate comparison
orig = df["default"].value_counts(normalize=True).rename("original")
samp = df_sample["default"].value_counts(normalize=True).rename("sample")
print(pd.concat([orig, samp], axis=1))

# check issue_d year distribution
df["issue_d"] = pd.to_datetime(df["issue_d"])
df_sample["issue_d"] = pd.to_datetime(df_sample["issue_d"])
print(df["issue_d"].dt.year.value_counts(normalize=True).sort_index().head())
print(df_sample["issue_d"].dt.year.value_counts(normalize=True).sort_index().head())
```
